```python
import numpy as np
import matplotlib.pyplot as plt

```

```python

X=np.array ([1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0])
y=np.array ([4.0, 4.55, 6.7, 7.25, 9.3, 9.45, 11.6, 11.85, 13.9, 14.65, 16.7, 17.05, 19.0, 19.45, 21.6])
def gradient_descent(X,y,w_init,b_init,lr,iteration):
  w=w_init
  b=b_init
  cost_history=[]
  for i in range(iteration):
    #hypothesis
    y_hat=X*w+b
    error=y_hat-y
    #cost things
    cost=np.mean(error**2)
    cost_history.append(cost)
    #gradient descent
    dw=np.mean(error*X)
    db=np.mean(error)
    #updated values
    w=w-lr*dw
    b=b-lr*db

  return w,b,cost_history

iteration=1000
w_init, b_init=-1.0, 100
#normal
w_good, b_good, cost_good = gradient_descent(X,y,w_init,b_init,lr=0.01,iteration=iteration)
#high
w_high, b_high, cost_high = gradient_descent(X,y,w_init,b_init,lr=0.15,iteration=iteration)
#low
w_low, b_low, cost_low = gradient_descent(X,y,w_init,b_init,lr=0.001,iteration=iteration)

X_b=np.c_[np.ones_like(X),X]

theta=np.linalg.inv(X_b.T@X_b)@X_b.T@y
exact_b=theta[0]
exact_w=theta[1]

print("--- Final Results ---")
print(f"Closed-form (Normal Equation): w = {exact_w:.4f}, b = {exact_b:.4f}")
print(f"Gradient Descent (Good LR)   : w = {w_good:.4f}, b = {b_good:.4f}")
print(f"Gradient Descent (High LR)   : w = {w_high:.4f}, b = {b_high:.4f}")
print(f"Gradient Descent (Low LR)    : w = {w_low:.4f}, b = {b_low:.4f}")


plt.plot(cost_good, label="lr=0.01")
plt.plot(cost_low, label="lr=0.001")
plt.plot(cost_high[:20], label="lr=0.15 (first 20 iters)")
plt.yscale("log"); plt.xlabel("iteration"); plt.ylabel("MSE"); plt.legend(); plt.show()

```
